```python
import torch
from transformers import AutoImageProcessor, AutoModel
from PIL import Image
import requests

url = 'http://images.cocodataset.org/val2017/000000039769.jpg'
image = Image.open(requests.get(url, stream=True).raw)
print(image.size)
processor = AutoImageProcessor.from_pretrained('/Users/tannicholas/Downloads/dinov2_hf')
print(processor)
model = AutoModel.from_pretrained('/Users/tannicholas/Downloads/dinov2_hf')

inputs = processor(images=image, return_tensors="pt")
print(torch.max(inputs.pixel_values) )
print(inputs.pixel_values.size())
outputs = model(**inputs)
last_hidden_states = outputs[0]
cls_token = last_hidden_states[:, 0, : ]
pooler_output = outputs[1]
# We have to force return_dict=False for tracing
model.config.return_dict = False

with torch.no_grad():
    traced_model = torch.jit.trace(model, [inputs.pixel_values])
    traced_outputs = traced_model(inputs.pixel_values)

print((last_hidden_states - traced_outputs[0]).abs().max())
print(last_hidden_states.size())
print(model)
#TODO: to pool or not to pool?
# print(outputs[1].size())
# print(pooler_output==cls_token)
```
